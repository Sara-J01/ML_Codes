```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Step 1: Create sample dataset
data = {
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Annual_Spend": [2000, 2500, 3000, 8000, 9000, 10000, 20000, 22000],
    "Visits": [2, 3, 4, 10, 11, 12, 20, 25]
}

df = pd.DataFrame(data)

print("Original Data")
print(df)

# Step 2: Select input features
X = df[["Annual_Spend", "Visits"]]

# Step 3: Standardize the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Create K-Means model
model = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Step 5: Perform clustering
df["Cluster"] = model.fit_predict(X_scaled)

print("\nClustered Data")
print(df)

# Step 6: Plot the clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Annual_Spend"],
    df["Visits"],
    c=df["Cluster"],
    s=100
)

plt.xlabel("Annual Spending")
plt.ylabel("Number of Visits")
plt.title("K-Means Customer Clustering")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering
from scipy.cluster.hierarchy import dendrogram, linkage

# Step 1: Create sample dataset
data = {
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Annual_Spend": [2000, 2500, 3000, 8000, 9000, 10000, 20000, 22000],
    "Visits": [2, 3, 4, 10, 11, 12, 20, 25]
}

df = pd.DataFrame(data)

print("Original Data")
print(df)

# Step 2: Select features
X = df[["Annual_Spend", "Visits"]]

# Step 3: Standardize
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 4: Create dendrogram
linked = linkage(X_scaled, method="ward")

plt.figure(figsize=(10, 6))

dendrogram(
    linked,
    labels=df["Customer"].values
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customer")
plt.ylabel("Distance")
plt.show()

# Step 5: Perform hierarchical clustering
model = AgglomerativeClustering(
    n_clusters=3,
    linkage="ward"
)

df["Cluster"] = model.fit_predict(X_scaled)

print("\nClustered Data")
print(df)

# Step 6: Visualize clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Annual_Spend"],
    df["Visits"],
    c=df["Cluster"],
    s=100
)

for i in range(len(df)):
    plt.text(
        df["Annual_Spend"][i],
        df["Visits"][i],
        df["Customer"][i]
    )

plt.xlabel("Annual Spending")
plt.ylabel("Number of Visits")
plt.title("Hierarchical Clustering")
plt.show()
```

```python
# Import libraries
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler

# Create sample data
data = {
    "Customer": ["A", "B", "C", "D", "E", "F", "G", "H"],
    "Annual_Spend": [2000, 2500, 3000, 8000, 9000, 10000, 20000, 22000],
    "Visits": [2, 3, 4, 10, 11, 12, 20, 25]
}

df = pd.DataFrame(data)

print("Original Data:")
print(df)

# Select features
X = df[["Annual_Spend", "Visits"]]

# Standardize the values
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Create K-Means model
kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

# Train the model and assign clusters
df["Cluster"] = kmeans.fit_predict(X_scaled)

print("\nClustered Data:")
print(df)

# Plot clusters
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Annual_Spend"],
    df["Visits"],
    c=df["Cluster"],
    s=120
)

# Add customer names to graph
for i in range(len(df)):
    plt.text(
        df["Annual_Spend"][i] + 200,
        df["Visits"][i],
        df["Customer"][i]
    )

plt.xlabel("Annual Spending")
plt.ylabel("Number of Visits")
plt.title("K-Means Customer Clustering")
plt.show()
```
